```python

# 01 - Pricing and Listing Visualizations

import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Load Data

listings = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\listings_featured.csv"
)


# Output Folder

image_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Visualizations\01_pricing_listing"

os.makedirs(image_path, exist_ok=True)


# 1. Price Distribution

plt.figure(figsize=(10, 6))

sns.histplot(
    listings["price"].dropna(),
    bins=50
)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "price_distribution.png"),
    bbox_inches="tight"
)

plt.close()


# 2. Log-Transformed Price Distribution

plt.figure(figsize=(10, 6))

sns.histplot(
    listings["price_log"].dropna(),
    bins=50
)

plt.title("Log-Transformed Price Distribution")
plt.xlabel("Log Price")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "log_price_distribution.png"),
    bbox_inches="tight"
)

plt.close()


# 3. Price by Room Type

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=listings,
    x="room_type",
    y="price"
)

plt.title("Price by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Price")

plt.savefig(
    os.path.join(image_path, "price_by_room_type.png"),
    bbox_inches="tight"
)

plt.close()


# 4. Top 10 Neighbourhoods by Median Price

top_neighbourhoods = (
    listings
    .dropna(subset=["price"])
    .groupby("neighbourhood")["price"]
    .median()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

top_neighbourhoods.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Neighbourhoods by Median Price")
plt.xlabel("Median Price")
plt.ylabel("Neighbourhood")

plt.savefig(
    os.path.join(image_path, "neighbourhood_price.png"),
    bbox_inches="tight"
)

plt.close()


# 5. Listings by Room Type

plt.figure(figsize=(8, 5))

sns.countplot(
    data=listings,
    x="room_type"
)

plt.title("Listings by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "listings_by_room_type.png"),
    bbox_inches="tight"
)

plt.close()


# 6. Listings by Neighbourhood Group

plt.figure(figsize=(8, 5))

sns.countplot(
    data=listings,
    x="neighbourhood_group"
)

plt.title("Listings by Neighbourhood Group")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "listings_by_neighbourhood_group.png"),
    bbox_inches="tight"
)

plt.close()


# 7. Listings by Price Category

plt.figure(figsize=(8, 5))

sns.countplot(
    data=listings,
    x="price_category"
)

plt.title("Listings by Price Category")
plt.xlabel("Price Category")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "listings_by_price_category.png"),
    bbox_inches="tight"
)

plt.close()


# 8. Listings by Host Type

plt.figure(figsize=(8, 5))

sns.countplot(
    data=listings,
    x="host_type"
)

plt.title("Listings by Host Type")
plt.xlabel("Host Type")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "listings_by_host_type.png"),
    bbox_inches="tight"
)

plt.close()


# 9. Listings by Availability Category

plt.figure(figsize=(8, 5))

sns.countplot(
    data=listings,
    x="availability_category"
)

plt.title("Listings by Availability Category")
plt.xlabel("Availability Category")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "listings_by_availability_category.png"),
    bbox_inches="tight"
)

plt.close()


# 10. Price vs Availability

price_data = listings[
    ["price", "availability_365"]
].dropna()

price_data = price_data.sample(
    min(5000, len(price_data)),
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=price_data,
    x="availability_365",
    y="price",
    alpha=0.4
)

plt.title("Price vs Availability")
plt.xlabel("Availability (Days)")
plt.ylabel("Price")

plt.savefig(
    os.path.join(image_path, "price_vs_availability.png"),
    bbox_inches="tight"
)

plt.close()


print("All pricing and listing visualizations saved successfully.")
print("Images saved to:")
print(image_path)

```
