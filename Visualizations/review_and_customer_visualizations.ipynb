```python



import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Load Data

listings = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\listings_featured.csv"
)

reviews = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\reviews_clean.csv"
)


# Output Folder

image_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Visualizations\02_review_customer"

os.makedirs(image_path, exist_ok=True)


# Convert Review Date

reviews["date"] = pd.to_datetime(
    reviews["date"],
    errors="coerce"
)


# 1. Reviews Over Time

reviews_by_year = (
    reviews
    .groupby(reviews["date"].dt.year)
    .size()
)

plt.figure(figsize=(10, 6))

reviews_by_year.plot()

plt.title("Reviews Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Reviews")

plt.savefig(
    os.path.join(image_path, "reviews_over_time.png"),
    bbox_inches="tight"
)

plt.close()


# 2. Reviews by Month

reviews_by_month = (
    reviews
    .groupby(reviews["date"].dt.month)
    .size()
)

plt.figure(figsize=(10, 6))

reviews_by_month.plot(kind="bar")

plt.title("Reviews by Month")
plt.xlabel("Month")
plt.ylabel("Number of Reviews")

plt.savefig(
    os.path.join(image_path, "reviews_by_month.png"),
    bbox_inches="tight"
)

plt.close()


# 3. Top 10 Reviewed Listings

top_listings = (
    reviews
    .groupby("listing_id")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

top_listings.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Reviewed Listings")
plt.xlabel("Number of Reviews")
plt.ylabel("Listing ID")

plt.savefig(
    os.path.join(image_path, "top_reviewed_listings.png"),
    bbox_inches="tight"
)

plt.close()


# 4. Top 10 Neighbourhoods by Review Activity

review_neighbourhood = (
    reviews
    .merge(
        listings[["id", "neighbourhood"]],
        left_on="listing_id",
        right_on="id",
        how="left"
    )
    .groupby("neighbourhood")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

review_neighbourhood.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Neighbourhoods by Review Activity")
plt.xlabel("Number of Reviews")
plt.ylabel("Neighbourhood")

plt.savefig(
    os.path.join(image_path, "reviews_by_neighbourhood.png"),
    bbox_inches="tight"
)

plt.close()


# 5. Reviews by Room Type

room_reviews = (
    listings
    .groupby("room_type")["number_of_reviews"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

room_reviews.plot(kind="bar")

plt.title("Reviews by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Number of Reviews")

plt.savefig(
    os.path.join(image_path, "reviews_by_room_type.png"),
    bbox_inches="tight"
)

plt.close()


# 6. Price vs Review Activity

price_reviews = listings[
    ["price", "number_of_reviews"]
].dropna()

price_reviews = price_reviews.sample(
    min(5000, len(price_reviews)),
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=price_reviews,
    x="price",
    y="number_of_reviews",
    alpha=0.4
)

plt.title("Price vs Review Activity")
plt.xlabel("Price")
plt.ylabel("Number of Reviews")

plt.savefig(
    os.path.join(image_path, "price_vs_reviews.png"),
    bbox_inches="tight"
)

plt.close()


# 7. Availability vs Review Activity

availability_reviews = listings[
    ["availability_365", "number_of_reviews"]
].dropna()

availability_reviews = availability_reviews.sample(
    min(5000, len(availability_reviews)),
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=availability_reviews,
    x="availability_365",
    y="number_of_reviews",
    alpha=0.4
)

plt.title("Availability vs Review Activity")
plt.xlabel("Availability (Days)")
plt.ylabel("Number of Reviews")

plt.savefig(
    os.path.join(image_path, "availability_vs_reviews.png"),
    bbox_inches="tight"
)

plt.close()


# 8. Reviews by Availability Category

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=listings,
    x="availability_category",
    y="number_of_reviews"
)

plt.title("Reviews by Availability Category")
plt.xlabel("Availability Category")
plt.ylabel("Number of Reviews")

plt.savefig(
    os.path.join(image_path, "reviews_by_availability_category.png"),
    bbox_inches="tight"
)

plt.close()


# 9. Listings With vs Without Reviews

review_status = (
    listings["has_reviews"]
    .value_counts()
)

plt.figure(figsize=(8, 5))

review_status.plot(kind="bar")

plt.title("Listings With vs Without Reviews")
plt.xlabel("Review Status")
plt.ylabel("Number of Listings")

plt.savefig(
    os.path.join(image_path, "listings_review_status.png"),
    bbox_inches="tight"
)

plt.close()


print("All review and customer visualizations saved successfully.")
print("Images saved to:")
print(image_path)
```
