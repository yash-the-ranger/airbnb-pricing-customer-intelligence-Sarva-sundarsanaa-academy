```python
import pandas as pd
import numpy as np

listings = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\listings_clean.csv"

)

reviews = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\reviews_clean.csv"

)
```

```python
print("Listings:", listings.shape)
print("Reviews:", reviews.shape)
```

Price_Category

```python
price_data = listings.dropna(subset=["price"])

print(price_data["price"].quantile([0.25, 0.50, 0.75]))
```

```python
listings["price_category"] = pd.cut(
    listings["price"],
    bins=[0, 98, 176, 303, np.inf],
    labels=["Budget", "Mid-Range", "Premium", "Luxury"],
    include_lowest=True
)
```

```python
print(listings["price_category"].value_counts())
```

```python
print(listings[["price", "price_category"]].head(10))
```

host_type

```python
listings["host_type"] = np.where(
    listings["calculated_host_listings_count"].isna(),
    "Unknown",
    np.where(
        listings["calculated_host_listings_count"] == 1,
        "Single-Listing Host",
        "Multi-Listing Host"
    )
)
```

```python
print(listings["host_type"].value_counts())
```

has_reviews

```python
listings["has_reviews"] = np.where(
    listings["number_of_reviews"] > 0,
    "Yes",
    "No"
)
```

```python
print(listings["has_reviews"].value_counts())
```

price_log

```python
listings["price_log"] = np.log1p(listings["price"])
```

```python
print(listings[["price", "price_log"]].head(10))
```

Availability category

```python
listings["availability_category"] = pd.cut(
    listings["availability_365"],
    bins=[-1, 120, 240, 365],
    labels=["Low", "Medium", "High"]
)
```

```python
print(listings["availability_category"].value_counts())
```
