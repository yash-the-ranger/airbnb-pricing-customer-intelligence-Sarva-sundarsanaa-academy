```python

# 04 - Review Analysis

import pandas as pd


# 1. Load Data

listings = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\listings_clean.csv"

reviews = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\reviews_clean.csv"
)


# 2. Convert Review Date

reviews["date"] = pd.to_datetime(
    reviews["date"],
    format="%d-%m-%Y",
    errors="coerce"
)


# 3. Reviews by Year

reviews["year"] = reviews["date"].dt.year

reviews_by_year = (
    reviews.groupby("year")
    .size()
    .reset_index(name="review_count")
)

print("\nReviews by Year:")
print(reviews_by_year)


# 4. Reviews by Month

reviews["month"] = reviews["date"].dt.month

reviews_by_month = (
    reviews.groupby("month")
    .size()
    .reset_index(name="review_count")
)

print("\nReviews by Month:")
print(reviews_by_month)


# 5. Top Reviewed Listings

top_listings = (
    reviews.groupby("listing_id")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="review_count")
)

print("\nTop 10 Reviewed Listings:")
print(top_listings)


# 6. Reviews by Neighbourhood

reviews_neighbourhood = (
    reviews
    .merge(
        listings[["id", "neighbourhood"]],
        left_on="listing_id",
        right_on="id",
        how="left"
    )
    .groupby("neighbourhood")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="review_count")
)

print("\nTop 10 Neighbourhoods by Review Activity:")
print(reviews_neighbourhood)


# 7. Reviews by Room Type

reviews_room_type = (
    reviews
    .merge(
        listings[["id", "room_type"]],
        left_on="listing_id",
        right_on="id",
        how="left"
    )
    .groupby("room_type")
    .size()
    .sort_values(ascending=False)
    .reset_index(name="review_count")
)

print("\nReviews by Room Type:")
print(reviews_room_type)


# 8. Reviews by Host Type

reviews_host_type = (
    reviews
    .merge(
        listings[["id", "host_type"]],
        left_on="listing_id",
        right_on="id",
        how="left"
    )
    .groupby("host_type")
    .size()
    .sort_values(ascending=False)
    .reset_index(name="review_count")
)

print("\nReviews by Host Type:")
print(reviews_host_type)


# 9. Price vs Review Activity

price_reviews = listings[
    ["price", "number_of_reviews", "room_type"]
].dropna(subset=["price"])

print("\nPrice vs Review Correlation:")

print(
    price_reviews[
        ["price", "number_of_reviews"]
    ].corr()
)


# 10. Recent Reviews

recent_reviews = (
    reviews
    .sort_values("date", ascending=False)
    [["listing_id", "date", "reviewer_name"]]
    .head(10)
)

print("\nRecent Reviews:")
print(recent_reviews)
```
