```python

# 02 Data Cleaning

import pandas as pd
import os


# 1. File Paths

listings_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\listings.csv"

reviews_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\reviews.csv"

processed_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed"

os.makedirs(processed_path, exist_ok=True)


# 2. Load Raw Data

listings_raw = pd.read_csv(listings_path)
reviews_raw = pd.read_csv(reviews_path)

print("Raw Listings Shape:", listings_raw.shape)
print("Raw Reviews Shape:", reviews_raw.shape)


# 3. Create Working Copies

listings_clean = listings_raw.copy()
reviews_clean = reviews_raw.copy()


# 4. Remove Unnecessary Column

if "license" in listings_clean.columns:
    listings_clean = listings_clean.drop(columns=["license"])


# 5. Remove Exact Duplicate Rows

listings_before = len(listings_clean)
reviews_before = len(reviews_clean)

listings_clean = listings_clean.drop_duplicates()
reviews_clean = reviews_clean.drop_duplicates()

print("\nDuplicate rows removed:")
print("Listings:", listings_before - len(listings_clean))
print("Reviews:", reviews_before - len(reviews_clean))


# 6. Handle Missing Listing Names

listings_clean["name"] = listings_clean["name"].fillna(
    "Unknown Listing"
)


# 7. Convert Integer Columns to Nullable Integers

listings_clean["host_id"] = listings_clean["host_id"].astype("Int64")

listings_clean["host_profile_id"] = (
    listings_clean["host_profile_id"].astype("Int64")
)

listings_clean["minimum_nights"] = (
    listings_clean["minimum_nights"].astype("Int64")
)

listings_clean["calculated_host_listings_count"] = (
    listings_clean["calculated_host_listings_count"].astype("Int64")
)


# 8. Clean Price

invalid_price = (
    listings_clean["price"] <= 0
)

print("\nInvalid price records:", invalid_price.sum())

valid_price = (
    listings_clean["price"].notna()
    & (listings_clean["price"] > 0)
)

print("Valid price records:", valid_price.sum())
print("Missing price records:", listings_clean["price"].isna().sum())


# 9. Clean Minimum Nights

invalid_minimum_nights = (
    listings_clean["minimum_nights"].notna()
    & (listings_clean["minimum_nights"] <= 0)
)

print(
    "\nInvalid minimum nights:",
    invalid_minimum_nights.sum()
)


# 10. Clean Availability

invalid_availability = (
    (listings_clean["availability_365"] < 0)
    | (listings_clean["availability_365"] > 365)
)

print(
    "Invalid availability records:",
    invalid_availability.sum()
)


# 11. Convert Listing Review Dates

listings_clean["last_review"] = pd.to_datetime(
    listings_clean["last_review"],
    format="%d-%m-%Y",
    errors="coerce"
)


# 12. Clean Reviews Date

reviews_clean["date"] = pd.to_datetime(
    reviews_clean["date"],
    format="%d-%m-%Y",
    errors="coerce"
)


# 13. Check Review Data

print("\nReview date missing after conversion:")
print(reviews_clean["date"].isna().sum())

print("\nReviewer name missing:")
print(reviews_clean["reviewer_name"].isna().sum())

print("\nReview comments missing:")
print(reviews_clean["comments"].isna().sum())


# 14. Review ID Decision




# 15. Final Cleaning Summary

print("\nFINAL CLEANING SUMMARY")

print("\nListings Shape:", listings_clean.shape)
print("Reviews Shape:", reviews_clean.shape)

print("\nListings Missing Values:")

print(
    listings_clean.isnull().sum()[
        listings_clean.isnull().sum() > 0
    ]
)

print("\nReviews Missing Values:")

print(
    reviews_clean.isnull().sum()[
        reviews_clean.isnull().sum() > 0
    ]
)

print("\nListings Duplicate Rows:")
print(listings_clean.duplicated().sum())

print("\nReviews Duplicate Rows:")
print(reviews_clean.duplicated().sum())


# 16. Save Cleaned Data

listings_output = os.path.join(
    processed_path,
    "listings_clean.csv"
)

reviews_output = os.path.join(
    processed_path,
    "reviews_clean.csv"
)

listings_clean.to_csv(
    listings_output,
    index=False
)

reviews_clean.to_csv(
    reviews_output,
    index=False
)

print("\nFILES SAVED")

print(listings_output)
print(reviews_output)
```
