```python

# 05 - EDA and Insights

import pandas as pd


# 1. Load Data

listings = pd.read_csv(
    r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\processed\listings_featured.csv"
)


# 2. Pricing by Room Type

room_type_pricing = (
    listings
    .dropna(subset=["price"])
    .groupby("room_type")["price"]
    .agg(["mean", "median"])
    .sort_values("median", ascending=False)
)

print("\nPricing by Room Type:")
print(room_type_pricing)


# 3. Neighbourhood Pricing

neighbourhood_pricing = (
    listings
    .dropna(subset=["price"])
    .groupby("neighbourhood")["price"]
    .agg(["mean", "median", "count"])
    .sort_values("median", ascending=False)
)

print("\nTop Neighbourhoods by Price:")
print(neighbourhood_pricing.head(10))


# 4. Review Activity by Room Type

room_type_reviews = (
    listings
    .groupby("room_type")["number_of_reviews"]
    .agg(["mean", "median", "sum"])
    .sort_values("mean", ascending=False)
)

print("\nReview Activity by Room Type:")
print(room_type_reviews)


# 5. Host Analysis

host_analysis = (
    listings
    .groupby("host_type")
    .agg(
        listings=("id", "count"),
        average_price=("price", "mean"),
        average_reviews=("number_of_reviews", "mean")
    )
)

print("\nHost Analysis:")
print(host_analysis)


# 6. Availability Analysis

availability_analysis = (
    listings
    .groupby("availability_category", observed=True)
    .agg(
        listings=("id", "count"),
        average_price=("price", "mean"),
        average_reviews=("number_of_reviews", "mean")
    )
)

print("\nAvailability Analysis:")
print(availability_analysis)


# 7. Correlation Analysis

correlation = listings[
    ["price", "number_of_reviews", "availability_365"]
].corr()

print("\nCorrelation Matrix:")
print(correlation)


# 8. Listings With and Without Reviews

review_status = (
    listings
    .groupby("has_reviews")
    .agg(
        listings=("id", "count"),
        average_price=("price", "mean"),
        average_availability=("availability_365", "mean")
    )
)

print("\nListings With and Without Reviews:")
print(review_status)
```
