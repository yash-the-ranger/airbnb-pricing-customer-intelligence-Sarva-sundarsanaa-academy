```python


# 01 - Data Understanding


import pandas as pd



# 1. File paths


listings_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\listings.csv"

reviews_path = r"C:\Users\yasht\Downloads\Airbnb revenue analytics\Datasets\reviews.csv"



# 2. load dataset


listings = pd.read_csv(listings_path)

reviews = pd.read_csv(reviews_path)



# 3.Basic info


print("=" * 60)
print("LISTINGS DATASET")
print("=" * 60)

print("Shape:", listings.shape)

print("\nColumns:")
print(listings.columns.tolist())

print("\nFirst 5 rows:")
print(listings.head())

print("\nData types and information:")
listings.info()


print("\n" + "=" * 60)
print("REVIEWS DATASET")
print("=" * 60)

print("Shape:", reviews.shape)

print("\nColumns:")
print(reviews.columns.tolist())

print("\nFirst 5 rows:")
print(reviews.head())

print("\nData types and information:")
reviews.info()



# 4.Missing values 


print("\n" + "=" * 60)
print("MISSING VALUES - LISTINGS")
print("=" * 60)

print(
    listings.isnull()
    .sum()
    .sort_values(ascending=False)
)


print("\n" + "=" * 60)
print("MISSING VALUES - REVIEWS")
print("=" * 60)

print(
    reviews.isnull()
    .sum()
    .sort_values(ascending=False)
)


# 5.Duplicate Rows


print("\n" + "=" * 60)
print("DUPLICATE ROWS")
print("=" * 60)

print(
    "Listings duplicate rows:",
    listings.duplicated().sum()
)

print(
    "Reviews duplicate rows:",
    reviews.duplicated().sum()
)



# 6.Duplicate Id's


print("\n" + "=" * 60)
print("DUPLICATE IDS")
print("=" * 60)

print(
    "Duplicate listing IDs:",
    listings["id"].duplicated().sum()
)

print(
    "Duplicate review IDs:",
    reviews["id"].duplicated().sum()
)



# 7. Unique values


print("\n" + "=" * 60)
print("UNIQUE VALUES")
print("=" * 60)

print(
    "Unique listings:",
    listings["id"].nunique()
)

print(
    "Unique hosts:",
    listings["host_id"].nunique()
)

print(
    "Unique reviewers:",
    reviews["reviewer_id"].nunique()
)

print(
    "Unique review IDs:",
    reviews["id"].nunique()
)



# 8.Room type distribution


print("\n" + "=" * 60)
print("ROOM TYPE DISTRIBUTION")
print("=" * 60)

print(
    listings["room_type"]
    .value_counts(dropna=False)
)



# 9. Neighbourhood Group Distribution


print("\n" + "=" * 60)
print("NEIGHBOURHOOD GROUP DISTRIBUTION")
print("=" * 60)

print(
    listings["neighbourhood_group"]
    .value_counts(dropna=False)
)



# 10.Top Neighbourhoods


print("\n" + "=" * 60)
print("TOP 15 NEIGHBOURHOODS")
print("=" * 60)

print(
    listings["neighbourhood"]
    .value_counts()
    .head(15)
)



# 11.Numerical summary


print("\n" + "=" * 60)
print("LISTINGS NUMERICAL SUMMARY")
print("=" * 60)

print(
    listings.describe()
)



# 12.price understanding


print("\n" + "=" * 60)
print("PRICE ANALYSIS")
print("=" * 60)

print(
    "Missing prices:",
    listings["price"].isnull().sum()
)

print(
    "Zero prices:",
    (listings["price"] == 0).sum()
)

print(
    "Negative prices:",
    (listings["price"] < 0).sum()
)

print("\nPrice statistics:")

print(
    listings["price"].describe()
)

print("\nPrice quantiles:")

print(
    listings["price"].quantile(
        [
            0,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
            1.00
        ]
    )
)



# 13.minimum nights


print("\n" + "=" * 60)
print("MINIMUM NIGHTS")
print("=" * 60)

print(
    listings["minimum_nights"].describe()
)

print(
    "\nMissing minimum nights:",
    listings["minimum_nights"].isnull().sum()
)



# 14.number of reviews


print("\n" + "=" * 60)
print("NUMBER OF REVIEWS")
print("=" * 60)

print(
    listings["number_of_reviews"].describe()
)



# 15. avaliability


print("\n" + "=" * 60)
print("AVAILABILITY")
print("=" * 60)

print(
    listings["availability_365"].describe()
)



# 16. Review date understanding

print("\n" + "=" * 60)
print("REVIEW DATE")
print("=" * 60)

print(
    reviews["date"].head(20)
)

print(
    "\nRaw earliest date:",
    reviews["date"].min()
)

print(
    "Raw latest date:",
    reviews["date"].max()
)



# 17. review information


print("\n" + "=" * 60)
print("REVIEW INFORMATION")
print("=" * 60)

print(
    "Missing reviewer names:",
    reviews["reviewer_name"].isnull().sum()
)

print(
    "Missing comments:",
    reviews["comments"].isnull().sum()
)



# 18.Duplicate lisings and reviews id 


print("\n" + "=" * 60)
print("LISTING + REVIEW ID DUPLICATES")
print("=" * 60)

print(
    reviews.duplicated(
        subset=["listing_id", "id"]
    ).sum()
)


# 19. overview


print("\n" + "=" * 60)
print("FINAL DATA UNDERSTANDING SUMMARY")
print("=" * 60)

print(
    "Listings dataset:",
    listings.shape
)

print(
    "Reviews dataset:",
    reviews.shape
)

print(
    "\nListings columns:",
    len(listings.columns)
)

print(
    "Reviews columns:",
    len(reviews.columns)
)

print("\nData understanding completed.")
```
